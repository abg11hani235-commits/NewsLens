## 2. Lead-3 Heuristic Baseline
News articles follow the inverted pyramid convention: the most crucial information is located in the opening sentences.

```python
lead3_out = lead_n_summarize(sample['article'], num_sentences=3)
print("--- Lead-3 Generated Summary ---")
print(lead3_out)
```

## 3. Graph-Based TextRank (PageRank) Algorithm
TextRank models the document as a graph $G = (V, E)$ where:
- Vertices $V$ are sentences.
- Edge weights $E_{ij}$ are cosine similarities between TF-IDF representations.
- The PageRank algorithm iteratively propagates centrality scores across nodes.
- Top-scoring sentences are extracted and re-sorted into original discourse order.

```python
textrank_out = textrank_summarize(sample['article'], num_sentences=3)
print("--- TextRank Generated Summary ---")
print(textrank_out)
```

## 4. Qualitative Comparison of Extractive Approaches
We observe that while Lead-3 captures the event introduction, TextRank connects the main event to key stakeholder statements later in the document.

```python
comparison_table = pd.DataFrame({
    "Method": ["Human Highlights", "Lead-3 Baseline", "TextRank Extractive"],
    "Summary Text": [sample['highlights'], lead3_out, textrank_out],
    "Word Count": [len(sample['highlights'].split()), len(lead3_out.split()), len(textrank_out.split())]
})
comparison_table
```
